## Automated Dataset Profiling

### Helper Function to Automate Profiling

```python
def analyze_file(file_path):

    # Load dataset
    df = pd.read_csv(file_path)

    filename = Path(file_path).name

    # Clean column names
    df.columns = df.columns.str.strip()

    # Dataset Overview
    rows = df.shape[0]
    columns = df.shape[1]

    # Label Distribution
    label_counts = df["Label"].value_counts().to_dict()

    # Missing Values
    missing_count = df.isnull().sum().sum()

    # Missing Rate
    missing_rate = missing_count / df.size * 100

    # Missing Rows
    missing_rows = df.isnull().any(axis=1).sum()

    # Duplicated Rows
    duplicate_count = df.duplicated().sum()

    # Duplicated Rate
    duplicate_rate = df.duplicated().mean() * 100

    # Infinite Values
    numeric_df = df.select_dtypes(include=np.number)
    infinite_count = np.isinf(numeric_df).sum().sum()

    return {
        "file": file_path,
        "rows": rows,
        "columns": columns,
        "duplicate_count": duplicate_count,
        "duplicate_rate": duplicate_rate,
        "missing_count": missing_count,
        "missing_rate": missing_rate,
        "missing_rows": missing_rows,
        "infinite_count": infinite_count,
        "label_counts": label_counts
    }
```

```python
monday_result = analyze_file(
    "../data/raw/cicids2017/Monday-WorkingHours.pcap_ISCX.csv"
)

monday_result
```

```python
tuesday_result = analyze_file(
    "../data/raw/cicids2017/Tuesday-WorkingHours.pcap_ISCX.csv"
)

tuesday_result
```

```python
from pathlib import Path

data_dir = Path("../data/raw/cicids2017")

csv_files = list(data_dir.glob("*.csv"))

results = []

for file in csv_files:
    result = analyze_file(file)
    results.append(result)

summary_df = pd.DataFrame(results)

summary_df

summary_df.columns

summary_df.shape
```

```python
summary_df[[
    "file",
    "rows",
    "columns",
    "duplicate_count",
    "duplicate_rate",
    "missing_count",
    "missing_rate",
    "missing_rows",
    "infinite_count"
]]

summary_df.loc[0, "label_counts"]

for label, count in summary_df.loc[0, "label_counts"].items():
    print(label, count)
```

```python
label_results = []

for _, row in summary_df.iterrows():
    for label, count in row["label_counts"].items():
        label_results.append({
            "file" : row["file"],
            "label" : label,
            "count" : count
        })

label_summary_df = pd.DataFrame(label_results)
label_summary_df
```

```python
label_summary_df = label_summary_df.merge(
    summary_df[["file", "rows"]],
    on="file",
    how="left"
)

label_summary_df
```

```python
label_summary_df["percentage"] = (
    label_summary_df["count"]
    / label_summary_df["rows"]
    * 100
)

label_summary_df
```

```python
label_totals = label_summary_df.groupby("file")["count"].sum()

row_totals = summary_df.set_index("file")["rows"]

print(label_totals.equals(row_totals))
```

```python
dataset_label_summary = (
    label_summary_df
    .groupby("label")["count"]
    .sum()
    .sort_values(ascending=False)
)

dataset_label_summary
```

```python
dataset_label_summary = dataset_label_summary.reset_index()
dataset_label_summary
```

```python
total_flows = dataset_label_summary["count"].sum()

dataset_label_summary["percentage"] = (
    dataset_label_summary["count"]
    / total_flows
    * 100
)

dataset_label_summary
```

## Dataset Overview Conclusions

The initial exploratory analysis provides the following observations about CIC-IDS2017:

1. The dataset consists of eight CSV files containing a common set of 78 numerical network-flow features and a `Label` column.

2. The individual files represent substantially different traffic distributions. Some captures contain only benign traffic, while others contain specfic attack categories such as DDoS, PortScan, Bot, Infiltration, Patator and DoS attacks. Therefore, analysis of a single capture file would not be representative of the complete dataset.

3. Across the eight files, 15 distinct labels were identified: one benign class and fourteen attack classes.

4. The dataset exhibits substantial class imbalance. BENIGN traffic represents approximately 80.3% of all observed flows, while several attack classes contain very few samples. Heartbleed, for example, contains only 11 flows.

5. Class distributions also vary considerably between individual capture files. Consequently, both dataset-wide and per-file class distributions are important when evaluating the dataset for machine-learning use.

6. Several data quality issues were identified, including duplicated rows, missing values and infinite numerical values. These issues will require further investigation before preprocessing and model training.

7. The automated profiling and label aggregation pipeline was validated by comparing aggregated label counts against the original row counts for each file. The validation succeeded for all eight files.

### Open Questions

The overview phase does not yet determine whether CIC-IDS2017 is suitable for the final machine-learning project. Further analysis is required to investigate:

- the distribution and severity of missing/infinite values across features;
- the extent and nature of duplicate records;
- feature distributions and potential outliers;
- constant or near-constant features;
- correlations and redundant features;
- the consistency of feature schemas across the dataset;
- and ultimately, how these characteristics compare with CIC-IDS2018 and CIC-DDoS2019.

The next phase will therefore focus on a feature-level profiling and data quality analysis.
